# 4. Methods and Plan

One method to address my question is KNN classification. This method was chosen because it deals with the non linear relationship of my data as and the interaction between age and experience of players. The model works by calculating the distance between a new player and all existing players , and then classifying the new player by a majority vote between the K closest neighbors. Some requirements for the data such as the continuous age variable and categorical expereince variable is that they must be standardized so that they each contribute equally to the distance calculations. The main limitation of the model is that it is very sensitive to the number of neighbors (K) and it would be very slow if the number of users grew by a lot. This method is appropraite because it is a binary classification problem. To process the data we will remove the rows with missing age data, split the data for training and testing data, the categorical experience variable must be converted into numerical data, and k-fold cross validation must be performed on the training data to determine the optimal value of K.
