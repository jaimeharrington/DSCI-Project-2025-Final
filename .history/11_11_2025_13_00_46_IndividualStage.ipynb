# 2. Question
### Broad Question: 
What player characteristics and behaviours are most predictive of subscribing to a game-related newsletter, and how do these features differ between various player types?
### Specific Question:
Can a players age and self reported experience level reliably predict whether they will be a subscriber in the dataset?

Response variable: subscribe

Explanatory variable: Age, experience

### How the Data Adresses Question of Interest 

The data will help me address my specific question as it will allow me to identify the overall individual effect of age and experience on whether or not a user is subscribed or not but it will also allow me to test whether there is a corrolation between a players age and experience and whether the interaction between these two variables has an effect on whether or not a used is subscribed.

# 3. Exploratory Data Analysis and Visualization

```r
players <- read.csv("players.csv")
players
```

```r
summarize(players,
          mean_age = mean(Age, na.rm = TRUE),
          mean_hours_played = mean(played_hours))
```

```r
age_vs.subscription_plot <- players |>
    ggplot(aes(x = subscribe, y = Age, colour = subscribe)) +
    geom_bar(stat = "identity") +
    labs(x = "Age of Player",
     y = "Subscription Status",
    colour = "Subscription Status")
    ggtitle("Age vs. Subscription Status")

age_vs.subscription_plot
```
